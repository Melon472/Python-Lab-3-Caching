Лабораторна робота №3

Мета: отримати базові знання і навички в створенні декораторів. 

Варіант: 3

Завдання: Створити власний декоратор, для кешування результатів функції - якщо 
певна функція вже запускалась с такими параметрами – не запускати 
функцію, а відобразити результати із кешу. Протестувати на 3-4 функціях з 
різною кількістю та типом параметрів. 

```python
def cache_result(func):
    cache = {}


    def wrapper(*args, **kwargs):
        key = (args, tuple(sorted(kwargs.items())))

        if key in cache:
            print("Результат отримано з кешу")
            return cache[key]

        print("Виконується обчислення")
        result = func(*args, **kwargs)
        cache[key] = result

        return result
    return wrapper
```

```python
@cache_result
def multiply(a, b):
    return a * b

@cache_result
def square(a):
    return a ** 2

@cache_result
def sum_three(a, b, c):
    return a + b + c

@cache_result
def repeat_text(text, count):
    return text * count
```

```python
print(multiply(3, 5))
print(multiply(3, 5))
print(multiply(4, 6))

print(square(5))
print(square(5))
print(square(8))

print(sum_three(13, 16, 18))
print(sum_three(13, 16, 18))
print(sum_three(10, 10, 10))

print(repeat_text("Python ", 3))
print(repeat_text("Python ", 3))
print(repeat_text("Hello ", 2))
```
